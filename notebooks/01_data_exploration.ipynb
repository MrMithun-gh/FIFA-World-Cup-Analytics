```python
import pandas as pd
import os

# Load all files
matches = pd.read_csv('../data/raw/matches_1930_2022.csv')
world_cup = pd.read_csv('../data/raw/world_cup.csv')
ranking_2022 = pd.read_csv('../data/raw/fifa_ranking_2022-10-06.csv')
ranking_2026 = pd.read_csv('../data/raw/fifa_ranking_2026-06-08.csv')
schedule = pd.read_csv('../data/raw/schedule_2026.csv')

# Quick overview of each file
for name, df in [('matches', matches), ('world_cup', world_cup), 
                  ('ranking_2022', ranking_2022), ('ranking_2026', ranking_2026),
                  ('schedule', schedule)]:
    print(f"\n{'='*50}")
    print(f"FILE: {name}")
    print(f"Shape: {df.shape}")
    print(f"Columns: {list(df.columns)}")
    print(f"Null counts:\n{df.isnull().sum()[df.isnull().sum() > 0]}")
    print(df.head(2))
```
